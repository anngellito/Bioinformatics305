```python
import numpy as np
import pandas as pd
```

```python
def tablichka(a):
    """
    это чтобы матрицы выглядели красивенько
    """
    df = pd.DataFrame(a, index=['A', 'T', 'G', 'C'])
    return df
```

```python
sites = ["GAGGTAAAC", "TCCGTAAGC", "CAGGTTGGA",
"ACAGTCAGC", "TAGGTCAGC", "CAGGTCAGC",
"CAGGTCGAT", "CAGGTCAGC", "CAGGTCAGC",
"CAGGTTGGC"]
```

## PFM

```python
pfm = np.zeros(shape=(4,len(sites[0])))
for site in sites:
    site = list(site)
    for j in range(len(site)):
        if site[j]=='A':
            pfm[0,j]+=1
        if site[j]=='T':
            pfm[1,j]+=1
        if site[j]=='G':
            pfm[2,j]+=1
        if site[j]=='C':
            pfm[3,j]+=1
```

```python
tablichka(pfm)
```

## PPM

```python
m = pfm + 0.1
ppm = m / np.sum(m, axis=0)
tablichka(ppm)
```

## PWM

```python
ppm_AT = np.log2(ppm[0:2] / 0.295)
ppm_GC = np.log2(ppm[2:4] / 0.205)

pwm = np.vstack((ppm_AT, ppm_GC))
tablichka(pwm)
```

## Максимальный и минимальный скор

```python
max_scores = np.max(pwm, axis=0)
min_scores = np.min(pwm, axis=0)
max_indices = np.argmax(pwm, axis=0)
min_indices = np.argmin(pwm, axis=0)

max_score = np.sum(max_scores)
min_score = np.sum(min_scores)
```

```python
nucl = ['A', 'T', 'G', 'C']
max_sequence = ''
min_sequence = ''
for i in max_indices:
    max_sequence += nucl[i]
for i in min_indices:
    min_sequence += nucl[i]
```

```python
print(f'Максмальный скор: {max_score}, последовательносьть: {max_sequence}')
print(f'Минимальный скор: {min_score}, последовательносьть: {min_sequence}')
```
